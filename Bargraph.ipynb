```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "city": ["Nairobi", "Mombasa", "Nakuru"],
    "amount": [3950, 2400, 950],
})

df.plot(kind="bar", x="city", y="amount")
plt.title("Total sales by city")
plt.show()
```
